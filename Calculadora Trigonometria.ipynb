# Trigonometria

## RELAÇÕES TRIGONOMÉTRICAS

Seja um triângulo retângulo ABC, de lados medindo a, b, e c e ângulos $\hat{A}$, $\hat{B}$ e $\hat{C}\mbox{ }$onde,
- Hipotenusa é o lado oposto ao ângulo reto;
- Cateto oposto é o lado oposto ao ângulo de interesse;
- Cateto adjacente lado que vizinho ao ângulo excluindo a hipotenusa.

São válidas as seguintes relações:


### Seno
$${\rm sen}=\dfrac{\mbox{cateto oposto}}{\rm hipotenusa}$$
### Cosseno
$${\rm cos}=\dfrac{\mbox{cateto adjacente}}{\rm hipotenusa}$$
### Tangente
$${\rm tg}=\dfrac{\mbox{cateto oposto}}{\mbox{cateto adjacente}}$$



## Calculadora Python

Execute os códigos a seguir.

```python
import math

# Entrada dos comprimentos dos lados
cateto_oposto = float(input("Digite o comprimento do cateto oposto: "))
cateto_adjacente = float(input("Digite o comprimento do cateto adjacente: "))
hipotenusa = float(input("Digite o comprimento da hipotenusa: "))

# Verificação se é triângulo retângulo (sem math.isclose)
if hipotenusa**2 == cateto_oposto**2 + cateto_adjacente**2:
    # Cálculos
    seno = cateto_oposto / hipotenusa
    cosseno = cateto_adjacente / hipotenusa
    tangente = cateto_oposto / cateto_adjacente

    # Saída
    print(f"Seno: {seno:.2f}")
    print(f"Cosseno: {cosseno:.2f}")
    print(f"Tangente: {tangente:.2f}")
    print("É um triângulo retângulo.")
else:
    print("Não é um triângulo retângulo.")

```
