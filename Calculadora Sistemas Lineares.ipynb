# Sistemas Lineares

## Definição
Uma **equação linear** (primeiro grau) é toda equação na forma:
$$ a_{1}x_{1} + a_{2}x_{2} + \cdots + a_{n}x_{n}= b $$
onde:
- $a_{1}, a_{2},…, a_{n}$ são coeficientes reais;
- $x_{1}, x_{2}, …, x_{n}$ são as incógnitas;
- $b$ é o termo independente.

## Sistemas de equações lineares

É um conjunto de $m$ equações lineares com $n$ incógnitas, conforme o representado adiante:


\begin{cases}
 a_{11}x_{1}+a_{12}x_{2}+a_{13}x_{3}+ \dots +a_{1n}x_{n}=b_{1}\\
 a_{21}x_{1}+a_{22}x_{2}+a_{23}x_{3}+ \dots +a_{2n}x_{n}=b_{2}\\
 \hspace{0.7cm}\vdots\\
 a_{m1}x_{1}+a_{m2}x_{2}+a_{m3}x_{3}+ \dots +a_{mn}x_{n}=b_{m}
\end{cases}

## Solução de um sistema de equações lineares

Os valores das variáveis que satisfazem simultaneamente a todas as equações de um sistema constituem sua solução. Esses valores são chamados de raízes do sistema de equações lineares. Para encontrar a solução de um sistema de equações lineares, aplicam-se os seguintes métodos:

### Método da substituição

Consiste em isolar uma das incógnitas em uma das equações e, então, substituir a expressão encontrada na outra equação.

Considere os $a$, $b$, $c$ e $d$ números reais e o sistema linear $2 \times 2$:

\begin{cases}
ax+by=e \hspace{2cm} (1)\\
cx+dy=f \hspace{2cm} (2)
\end{cases}

Isolando $y$ na equação $(2)$ têm-se:
$$y = \frac{f-cx}{d}.$$
Substituindo em $(1)$ resulta numa equação que contém apenas a incógnita $x$ tornando possível descubrir o seu valor.
$$ax + b\left(\frac{f-cx}{d}\right) = e.$$

Para sistemas lineres com 3 incógnitas repetimos o procedimento mais uma vez até restar apenas uma incógnita a descobrir o valor.

### Método por Eliminação ou Adição

No método de eliminação, também conhecido como método de eliminação de Gauss ou método de escalonamento, o objetivo é transformar o sistema em sua forma mais simples (escalonada), de modo que seja possível resolver as incógnitas por substituição regressiva.

Considere os sistemas de equações lineares:

\begin{cases}
a_{11}x+a_{12}y=b_{1}\\
a_{21}x+a_{22}y=b_{2}
\end{cases}

ou

\begin{cases}
a_{11}x+a_{12}y+a_{13}z=b_{1}\\
a_{21}x+a_{22}y+a_{23}z=b_{2}\\
a_{31}x+a_{32}y+a_{33}z=b_{3}
\end{cases}

Resolver um sistema linear significa encontrar todos os valores que satisfazem as equações do sistema. Para isso, usamos algumas operações simples: podemos trocar duas linhas de lugar, multiplicar uma linha por um número diferente de zero ou somar/subtrair uma linha da outra. Essas operações servem para deixar a matriz do sistema em uma forma mais simples, chamada triangular, o que facilita bastante a resolução passo a passo das variáveis.

### Método de Cramer

Os sistemas de equações lineares

$$
\begin{cases}
a_{1}x + b_{1}y = c_{1} \\
a_{2}x + b_{2}y = c_{2}
\end{cases}
$$

e

$$
\begin{cases}
a_{1}x + b_{1}y + c_{1}z = d_{1} \\
a_{2}x + b_{2}y + c_{2}z = d_{2} \\
a_{3}x + b_{3}y + c_{3}z = d_{3}
\end{cases}
$$

possuem as respectivas matrizes dos coeficientes:

$$
\begin{bmatrix}
a_{1} & b_{1} \\
a_{2} & b_{2}
\end{bmatrix}
\quad \text{e} \quad
\begin{bmatrix}
a_{1} & b_{1} & c_{1} \\
a_{2} & b_{2} & c_{2} \\
a_{3} & b_{3} & c_{3}
\end{bmatrix}
$$

Tais matrizes possuem os seguintes determinantes:

\begin{align}
D_2 &= a_{1}b_{2} - a_{2}b_{1}; \\
D_3 &= a_{1}b_{2}c_{3} + b_{1}c_{2}a_{3} + c_{1}a_{2}b_{3}
\\
& - a_{3}b_{2}c_{1} - b_{3}c_{2}a_{1} - c_{3}a_{2}b_{1};
\end{align}

Pelo método de Cramer, temos:

$$
x = \frac{D_{x}}{D}, \quad y = \frac{D_{y}}{D}, \quad z = \frac{D_{z}}{D},
$$

em que os determinantes $D_{x}$, $D_{y}$ e $D_{z}$ são obtidos substituindo-se, respectivamente, as colunas correspondentes às variáveis $x$, $y$ e $z$ pela coluna dos termos independentes (resultados), ou seja, $c_{1}, c_{2}$ para o sistema $2 \times 2$, e $d_{1}, d_{2}, d_{3}$ para o sistema $3 \times 3$.

#### Condições de Solução

Seja o sistema de equações lineares:

$$
\begin{cases}
a_{1}x + b_{1}y = c_{1} \\
a_{2}x + b_{2}y = c_{2}
\end{cases}
$$

Quanto ao conjunto solução, existem as seguintes possibilidades, estabelecidas pelas relações abaixo descritas:

- *Sistema possível e determinado (solução única)*

$$
\dfrac{a_{1}}{a_{2}} \neq \dfrac{b_{1}}{b_{2}}
$$

- *Sistema possível e com infinitas soluções*

$$
\dfrac{a_{1}}{a_{2}} = \dfrac{b_{1}}{b_{2}} = \dfrac{c_{1}}{c_{2}}
$$

- *Sistema impossível (sem solução)*

$$
\dfrac{a_{1}}{a_{2}} = \dfrac{b_{1}}{b_{2}} \neq \dfrac{c_{1}}{c_{2}}
$$








# Calculadora Python

Execute os códigos a seguir.

```python
print("=== Calculadora de Sistemas Lineares 2x2 ===")

# Entrada dos coeficientes
a1 = float(input("Digite a1: "))
b1 = float(input("Digite b1: "))
c1 = float(input("Digite c1: "))

a2 = float(input("Digite a2: "))
b2 = float(input("Digite b2: "))
c2 = float(input("Digite c2: "))

# Cálculo do determinante
det = a1 * b2 - a2 * b1

if det == 0:
    print("O sistema não tem solução única (possivelmente impossível ou indeterminado).")
else:
    # Regra de Cramer
    x = (c1 * b2 - c2 * b1) / det
    y = (a1 * c2 - a2 * c1) / det

    print("\nSolução:")
    print(f"x = {x}")
    print(f"y = {y}")
```

```python
print("=== Resolver Sistema Linear 3x3 ===")

# Entrada dos coeficientes
a1 = float(input("Digite a1: "))
b1 = float(input("Digite b1: "))
c1 = float(input("Digite c1: "))
d1 = float(input("Digite d1: "))

a2 = float(input("Digite a2: "))
b2 = float(input("Digite b2: "))
c2 = float(input("Digite c2: "))
d2 = float(input("Digite d2: "))

a3 = float(input("Digite a3: "))
b3 = float(input("Digite b3: "))
c3 = float(input("Digite c3: "))
d3 = float(input("Digite d3: "))

# Determinante principal (D)
D = (a1 * (b2 * c3 - b3 * c2)
   - b1 * (a2 * c3 - a3 * c2)
   + c1 * (a2 * b3 - a3 * b2))

if D == 0:
    print("\nO sistema não tem solução única (determinante = 0).")
else:
    # Dx
    Dx = (d1 * (b2 * c3 - b3 * c2)
        - b1 * (d2 * c3 - d3 * c2)
        + c1 * (d2 * b3 - d3 * b2))

    # Dy
    Dy = (a1 * (d2 * c3 - d3 * c2)
        - d1 * (a2 * c3 - a3 * c2)
        + c1 * (a2 * d3 - a3 * d2))

    # Dz
    Dz = (a1 * (b2 * d3 - b3 * d2)
        - b1 * (a2 * d3 - a3 * d2)
        + d1 * (a2 * b3 - a3 * b2))

    x = Dx / D
    y = Dy / D
    z = Dz / D

    print("\nSolução do sistema:")
    print(f"x = {x}")
    print(f"y = {y}")
    print(f"z = {z}")
```
